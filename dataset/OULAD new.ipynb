# Step 7: Clean studentRegistration Dataset

## Handle Missing Values

The `studentRegistration` dataset contains missing values in two columns. The `date_registration` column has only 45 missing values, while the `date_unregistration` column has a large number of missing values. Missing values in `date_unregistration` indicate that the student did not unregister from the course, so they carry meaningful information and should not be removed.

```python
studentRegistration.isnull().sum()
```

```python
studentRegistration["date_registration"] = studentRegistration["date_registration"].fillna(
    studentRegistration["date_registration"].median()
)
```

```python
studentRegistration.isnull().sum()
```

## Observation

The missing values in `date_registration` were replaced using the median value because only a small number of records (45) were missing. The missing values in `date_unregistration` were retained because they indicate that a student did not withdraw from the course, making them meaningful rather than erroneous.

```python
studentRegistration["is_withdrawn"] = studentRegistration["date_unregistration"].notna().astype(int)
```

# Step 10: Clean and Aggregate studentAssessment Dataset

### Explanation

The `studentAssessment` dataset contains multiple assessment records for each student. Before aggregation, missing values in the `score` column are removed because the score is the most important feature for measuring academic performance.

```python
studentAssessment = studentAssessment.dropna(subset=["score"])
```

### Explanation

The missing values in the `score` column were removed because only 173 records were incomplete, representing less than 1% of the dataset.

```python
studentAssessment.isnull().sum()
```

### Explanation

The `studentAssessment` dataset is merged with the `assessments` dataset using the `id_assessment` column. This adds assessment information such as module, presentation, assessment type, and weight to each student's assessment record.

```python
studentAssessmentMerged = pd.merge(
    studentAssessment,
    assessments,
    on="id_assessment",
    how="left"
)
```

```python
studentAssessmentMerged.head()
```

### Explanation

The assessment records are grouped by student, module, and presentation to create one record per student. Features such as average score, maximum score, minimum score, and total assessments are calculated.

```python
assessment_summary = studentAssessmentMerged.groupby(
    ["id_student", "code_module", "code_presentation"]
).agg(
    average_score=("score", "mean"),
    max_score=("score", "max"),
    min_score=("score", "min"),
    total_assessments=("score", "count")
).reset_index()
```

```python
assessment_summary.head()
```

```python
assessment_summary.shape
```

## Observation

The assessment records were successfully aggregated into student-level features. Each row now represents one student enrolled in a specific module and presentation. These aggregated features will be merged with the main dataset in later steps.

## Observation

The assessment records were successfully aggregated into student-level features. The original dataset contained 173,912 assessment records, which were reduced to 25,820 student-module records after aggregation. The generated features include average score, maximum score, minimum score, and total number of assessments. These features will be used for machine learning model training.

# Step 12: Aggregate studentVle Dataset

### Explanation

The `studentVle` dataset contains more than 10 million interaction records. Instead of merging the raw data, it is aggregated to generate student-level engagement features. This reduces the dataset size and makes it suitable for machine learning.

```python
vle_summary = studentVle.groupby(
    ["id_student", "code_module", "code_presentation"]
).agg(
    total_clicks=("sum_click", "sum"),
    average_clicks=("sum_click", "mean"),
    max_clicks=("sum_click", "max"),
    total_visits=("sum_click", "count")
).reset_index()
```

The interaction records are grouped by student, module, and presentation. Four engagement features are created:
- **total_clicks:** Total clicks made by the student.
- **average_clicks:** Average clicks per interaction.
- **max_clicks:** Maximum clicks recorded in a single interaction.
- **total_visits:** Total number of interaction records.

```python
vle_summary.head()
```

```python
vle_summary.shape
```

## Observation

The `studentVle` dataset was successfully aggregated into student-level engagement features. The resulting dataset contains one record for each student-module-presentation combination and is ready to be merged with the assessment and student information datasets.

# Step 12B: Merge studentVle with VLE Dataset

### Explanation

The `studentVle` dataset contains student interaction logs, while the `vle` dataset contains information about each learning resource. These datasets are merged using `id_site`, `code_module`, and `code_presentation` to associate each student interaction with its corresponding activity type.

```python
studentVleMerged = pd.merge(
    studentVle,
    vle,
    on=["id_site", "code_module", "code_presentation"],
    how="left"
)
```

# Step 12C: Generate Student Engagement Features

The merged VLE dataset contains different types of learning activities such as forums, quizzes, homepage visits, and learning resources. These interactions are aggregated to generate engagement-based features for each student.

```python
vle_summary = studentVleMerged.groupby(
    ["id_student", "code_module", "code_presentation"]
).agg(
    total_clicks=("sum_click", "sum"),
    average_clicks=("sum_click", "mean"),
    max_clicks=("sum_click", "max"),
    total_visits=("sum_click", "count")
).reset_index()
```

```python
activity_summary = studentVleMerged.pivot_table(
    index=["id_student", "code_module", "code_presentation"],
    columns="activity_type",
    values="sum_click",
    aggfunc="sum",
    fill_value=0
).reset_index()
```

```python
activity_summary.head()
```

```python
activity_summary.shape
```

# Step 13: Merge All OULAD Datasets

### Explanation

The studentInfo dataset is used as the base dataset because it contains one record for each student enrolled in a module. The remaining datasets are merged using left joins to preserve all students while enriching the dataset with registration, assessment, engagement, activity, and course information.

```python
final_dataset = pd.merge(
    studentInfo,
    studentRegistration,
    on=["id_student", "code_module", "code_presentation"],
    how="left"
)
```

```python
final_dataset = pd.merge(
    final_dataset,
    assessment_summary,
    on=["id_student", "code_module", "code_presentation"],
    how="left"
)
```

```python
final_dataset = pd.merge(
    final_dataset,
    vle_summary,
    on=["id_student", "code_module", "code_presentation"],
    how="left"
)
```

```python
final_dataset = pd.merge(
    final_dataset,
    activity_summary,
    on=["id_student", "code_module", "code_presentation"],
    how="left"
)
```

```python
final_dataset = pd.merge(
    final_dataset,
    courses,
    on=["code_module", "code_presentation"],
    how="left"
)
```

All seven OULAD datasets were successfully merged into a single integrated dataset. Each row represents one student enrolled in a specific module and presentation. # Step 14: Handle Missing Values

The merged dataset contains missing values because some students did not attempt assessments or did not interact with the Virtual Learning Environment. These missing values are replaced with zero because the absence of activity indicates no participation rather than missing information.

```python
assessment_cols = [
    "average_score",
    "max_score",
    "min_score",
    "total_assessments"
]

final_dataset[assessment_cols] = final_dataset[assessment_cols].fillna(0)
```
